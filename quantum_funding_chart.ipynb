# Quantum Funding Geographic Bubble Chart — 2026 Update

This notebook generates a geographic bubble chart of public quantum-technology funding, styled after the original *Economist* “No small effort” graphic from the 2017 Technology Quarterly.

Data is drawn from the project CSV (`assets/quantum_funding_updated.csv`). Bubble area scales with annual or annualized figures. Positions are approximate geographic locations. China is treated as a special exception.

## 1. Imports and Data Loading

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Polygon
import numpy as np

df = pd.read_csv("assets/quantum_funding_updated.csv")
df.head()
```

## 2. Configuration

Define the values used for bubble sizing and the approximate geographic coordinates.

```python
values = {
    "United States": 1006,
    "Canada": 38.5,
    "Brazil": 5.5,
    "United Kingdom": 330,
    "Germany": 848,
    "Netherlands": 85,
    "Switzerland": 23,
    "Austria": 24,
    "Italy": 36,
    "India": 90,
    "South Korea": 140,
    "Japan": 241,
    "Singapore": 44,
    "Australia": None,
    "China": 15000,
}

pos = {
    "Canada": (-105, 57),
    "United States": (-99, 37),
    "Brazil": (-52, -10),
    "United Kingdom": (-3, 54),
    "Netherlands": (5, 53),
    "Germany": (11, 51),
    "Switzerland": (8, 47),
    "Austria": (14, 47),
    "Italy": (12, 42),
    "India": (79, 22),
    "South Korea": (128, 36),
    "Japan": (139, 37),
    "Singapore": (104, 1.3),
    "Australia": (134, -25),
    "China": (105, 35),
}
```

## 3. Generate the Chart

```python
fig, ax = plt.subplots(figsize=(16, 9), dpi=150)
fig.patch.set_facecolor("#fbfaf6")
ax.set_facecolor("#fbfaf6")

# Grid and continents (simplified)
for x in [-120, -60, 0, 60, 120]:
    ax.axvline(x, color="#e8e4db", lw=0.8, zorder=0)
for y in [-30, 0, 30, 60]:
    ax.axhline(y, color="#e8e4db", lw=0.8, zorder=0)

continents = [
    [(-165, 72), (-135, 72), (-110, 65), (-95, 55), (-82, 50), (-75, 35), (-100, 10), (-120, 20), (-145, 45)],
    [(-80, 12), (-50, 8), (-35, -10), (-45, -35), (-60, -55), (-72, -25)],
    [(-10, 70), (35, 70), (65, 60), (100, 55), (140, 50), (160, 35), (130, 5), (100, 10), (70, 25), (45, 35), (25, 45), (5, 40), (-10, 55)],
    [(112, -12), (153, -15), (154, -38), (130, -45), (114, -30)],
]
for coords in continents:
    ax.add_patch(Polygon(coords, closed=True, facecolor="#eeece5", edgecolor="#d8d4ca", linewidth=1.0, zorder=1))

# Bubbles
max_v = max(v for v in values.values() if v is not None and v < 5000)
for country, v in values.items():
    x, y = pos[country]
    if v is None:
        ax.scatter([x], [y], s=120, marker="o", facecolors="none", edgecolors="#6b7280", linewidths=1.8, zorder=4)
        continue
    if country == "China"::
        radius, color, edge, alpha = 8.5, "#f6e8d4", "#b96b26", 0.85
    else:
        radius = 2.2 + 9.5 * np.sqrt(v / max_v)
        color, edge, alpha = "#69c7d7", "#00a9c0", 0.72
    ax.add_patch(Circle((x, y), radius=radius, facecolor=color, edgecolor=edge, linewidth=1.6, alpha=alpha, zorder=3))

    if v >= 200 and country != "China"::
        ax.text(x, y, f"{v:,.0f}", ha="center", va="center", fontsize=10.5, color="#17333a", weight="bold", zorder=5)
        ax.text(x, y + 3.8, country, ha="center", va="center", fontsize=9, color="#17333a", weight="bold", zorder=6)
    elif country == "China"::
        ax.text(x, y, "≈15,000", ha="center", va="center", fontsize=11, color="#7d4519", weight="bold", zorder=5)
        ax.text(x, y + 5.5, "China\n(cumulative)", ha="center", va="center", fontsize=9, color="#7d4519", weight="bold", zorder=6)

fig.text(0.04, 0.94, "No small effort — 2026 update", fontsize=26, weight="bold", color="#282b2d")
fig.text(0.04, 0.895, "Estimated annual or annualized public quantum-technology funding (USD millions)", fontsize=13, color="#34383b")

ax.set_xlim(-175, 185)
ax.set_ylim(-60, 85)
ax.axis("off")
plt.tight_layout()
plt.show()
```

## 4. Notes

- U.S. figure is FY2024 enacted.
- Multi-year programs are annualized estimates.
- China is a cumulative estimate with high uncertainty and is not on the same scale.
